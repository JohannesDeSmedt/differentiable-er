```python
import pandas as pd
import numpy as np

df = pd.read_csv('results_nap_prediction_bpi12.csv')
df.tail()
```

```python
df = df[df['seed'] != 56]
pivot_table = pd.pivot_table(df, 
                             values=['f1', 'precision', 'accuracy'], 
                             
                             index=['model',  'd_model', 'lambda'], 
                             aggfunc=np.mean)
styled_table = pivot_table.style.background_gradient(cmap='Greens')
styled_table
```
